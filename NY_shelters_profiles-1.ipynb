```python
# delete any column that have all Nan value
df.dropna(axis=1, how='all', inplace=True)
df.head()
```

```python
# explode the 'latestPosts' column so that each post gets its own row
df_flat = df.explode('latestPosts').reset_index(drop=True)
df_flat.head()
```

```python
df_flat['post_timestamp'] = df_flat['latestPosts'].apply(lambda x: x.get('timestamp') if isinstance(x, dict) else None)
df_flat['post_likes'] = df_flat['latestPosts'].apply(lambda x: x.get('likesCount') if isinstance(x, dict) else None)
df_flat['post_comments'] = df_flat['latestPosts'].apply(lambda x: x.get('commentsCount') if isinstance(x, dict) else None)

print(f"a total of {df_flat.shape[0]} posts")
df_flat[['username', 'followersCount', 'post_timestamp', 'post_likes']].head(10)
```

```python
import numpy as np
df_flat['post_date'] = pd.to_datetime(df_flat['post_date'])

ig_profile_2 = df_flat.groupby('username').agg(
    nearest_post_date = ('post_date', 'max'),    # 最近的帖子日期
    earliest_post_date = ('post_date', 'min'),   # 最早的帖子日期
    total_number_of_posts = ('post_date', 'count'), # 最新捕获的帖子总数
    avg_likes=('post_likes', 'mean'),           # 近期平均点赞
    avg_comments=('post_comments', 'mean'),       # 近期平均评论
    followers_count=('followersCount', 'first')   # 粉丝基本盘
).reset_index()
```

```python
ig_profile_2['captured_days_span'] = (ig_profile_2['nearest_post_date'] - ig_profile_2['earliest_post_date']).dt.days + 1

ig_profile_2['posting_frequency'] = ig_profile_2['captured_days_span'] / ig_profile_2['total_number_of_posts']

safe_line = pd.to_datetime('2025-01-01')
ig_profile_2['Is_Suspicious_Recency'] = ig_profile_2['nearest_post_date'] < safe_line
```

```python
ig_profile_2['average_ER'] = ((ig_profile_2['avg_likes'] + ig_profile_2['avg_comments']) / (ig_profile_2['followers_count'] + 0.001)) * 100

# print(f"{ig_profile_2['average_ER'].mean()}")
print(f'{len(ig_profile_2)}')
```

```python
suspicious_ones = ig_profile_2[ig_profile_2['Is_Suspicious_Recency'] == True]
print(f"成功拦截到最新发帖日期异常的账号共 {len(suspicious_ones)} 个：")
print(suspicious_ones['username'].tolist())
```

```python
# 1. 筛选出最新帖子在2025年之后的健康账号
df_healthy = ig_profile_2[ig_profile_2['Is_Suspicious_Recency'] == False].copy()

# 2. 选择核心营销指标列
export_cols = ['username', 'followers_count', 'posting_frequency', 'average_ER', 'avg_likes', 'avg_comments']

# 3. 导出为 CSV 文件
df_healthy[export_cols].to_csv('NYS_Shelters_Healthy_Insta_Profiles.csv', index=False)
```

```python
df_healthy_real = df_healthy.dropna(subset=['posting_frequency'])

df_healthy_real[export_cols].to_csv(
    'NYS_Shelters_Healthy_real.csv',
    index=False
)
```

```python
# 计算粉丝数与发帖频率的皮尔逊相关系数（接近-1或1代表强相关，接近0代表无相关）
does_fans_number_correlate_with_freq = df_healthy['followers_count'].corr(df_healthy['posting_frequency'])

print("fans number vs frequency correlation:", does_fans_number_correlate_with_freq)
```

```python
# 计算发帖频率与平均互动率的相关系数
correlation_freq_er = df_healthy['posting_frequency'].corr(df_healthy['average_ER'])
print("发帖频率与互动率的相关系数:", correlation_freq_er)
```

```python
# 找出同时满足：发帖频率高于中位数 且 互动率高于中位数的账号
median_freq = df_healthy['posting_frequency'].median()
median_er = df_healthy['average_ER'].median()

# 筛选矩阵右上角（高频高ER）的明星账号
star_accounts = df_healthy[(df_healthy['posting_frequency'] <= median_freq) & (df_healthy['average_ER'] >= median_er)]
print("明星账号列表:")
print(star_accounts['username'].tolist())
```
